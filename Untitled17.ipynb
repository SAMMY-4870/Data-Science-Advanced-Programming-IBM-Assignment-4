```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, roc_auc_score, confusion_matrix,
    classification_report
)

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.neural_network import MLPClassifier

import warnings
warnings.filterwarnings("ignore")
```

```python

from google.colab import files
import os
import zipfile
import glob
import pandas as pd
import numpy as np

uploaded = files.upload()

csv_paths = []

for filename in uploaded.keys():
    if filename.lower().endswith(".csv"):
        csv_paths.append(filename)

    elif filename.lower().endswith(".zip"):
        with zipfile.ZipFile(filename, "r") as z:
            z.extractall("/content/hr_dataset")

        csv_paths.extend(
            glob.glob("/content/hr_dataset/**/*.csv", recursive=True)
        )

if not csv_paths:
    raise FileNotFoundError(
        "No CSV found. Upload the dataset CSV or its ZIP file."
    )

print("CSV files found:")
for i, path in enumerate(csv_paths):
    print(i, path)
```

```python

matches = [
    p for p in csv_paths
    if "hr-employee-attrition" in os.path.basename(p).lower()
]

DATA_PATH = matches[0] if matches else csv_paths[0]

df = pd.read_csv(DATA_PATH)
df.columns = df.columns.str.strip()

if "Attrition" not in df.columns:
    raise ValueError(
        f"'Attrition' column not found. Columns: {df.columns.tolist()}"
    )

df = df.drop_duplicates().copy()
df = df.dropna(subset=["Attrition"]).copy()
df["Attrition"] = (
    df["Attrition"].astype(str).str.strip().str.capitalize()
)

df = df[df["Attrition"].isin(["Yes", "No"])].copy()
df["Attrition"] = df["Attrition"].map({"No": 0, "Yes": 1}).astype(int)

print("Dataset:", DATA_PATH)
print("Shape:", df.shape)
print("Attrition counts:")
print(df["Attrition"].value_counts())
display(df.head())

assert df["Attrition"].nunique() == 2, "Target must contain both classes."
```

```python

import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display

sns.set_theme(style="whitegrid")

print("Dataset shape:", df.shape)
print("\nData types:")
display(df.dtypes.rename("Data Type").to_frame())

print("\nMissing values:")
display(df.isnull().sum().sort_values(ascending=False).head(15))

print("\nNumerical summary:")
display(df.describe().T)

attrition_rate = df["Attrition"].mean() * 100
print(f"\nOverall attrition rate: {attrition_rate:.2f}%")

plt.figure(figsize=(6, 4))
sns.countplot(data=df, x="Attrition")
plt.title("Employee Attrition Distribution")
plt.xlabel("Attrition (0 = Stayed, 1 = Left)")
plt.tight_layout()
plt.show()

if "Department" in df.columns:
    dept_rate = (
        df.groupby("Department")["Attrition"]
        .mean().mul(100).sort_values(ascending=False)
    )
    display(dept_rate.rename("Attrition Rate (%)").to_frame())

    plt.figure(figsize=(8, 4))
    sns.barplot(x=dept_rate.index, y=dept_rate.values)
    plt.ylabel("Attrition Rate (%)")
    plt.title("Attrition Rate by Department")
    plt.xticks(rotation=20)
    plt.tight_layout()
    plt.show()

if "Gender" in df.columns:
    gender_rate = df.groupby("Gender")["Attrition"].mean() * 100
    display(gender_rate.rename("Attrition Rate (%)").to_frame())
```

```python

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# Remove columns that contain no useful variation
constant_cols = [
    c for c in df.columns
    if c != "Attrition" and df[c].nunique(dropna=False) <= 1
]

X_raw = df.drop(columns=["Attrition"] + constant_cols)
y = df["Attrition"]

X_train_raw, X_test_raw, y_train, y_test = train_test_split(
    X_raw, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# One-hot encode categorical variables consistently
X_train = pd.get_dummies(
    X_train_raw, drop_first=True, dtype=int
)
X_test = pd.get_dummies(
    X_test_raw, drop_first=True, dtype=int
)

# Match test features to training features
X_test = X_test.reindex(columns=X_train.columns, fill_value=0)

# Fill any remaining missing feature values using training medians
medians = X_train.median(numeric_only=True)
X_train = X_train.fillna(medians).fillna(0)
X_test = X_test.fillna(medians).fillna(0)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
print("Number of features:", X_train.shape[1])
print("Training attrition rate:", round(y_train.mean() * 100, 2), "%")
print("Testing attrition rate:", round(y_test.mean() * 100, 2), "%")

```

```python

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.neural_network import MLPClassifier

from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, roc_auc_score, classification_report,
    confusion_matrix
)

models = {
    "Logistic Regression": (
        LogisticRegression(
            class_weight="balanced", max_iter=2000,
            random_state=42
        ), True
    ),
    "Random Forest": (
        RandomForestClassifier(
            n_estimators=200, class_weight="balanced",
            random_state=42, n_jobs=-1
        ), False
    ),
    "Gaussian Naive Bayes": (GaussianNB(), True),
    "Linear Discriminant Analysis": (
        LinearDiscriminantAnalysis(), True
    ),
    "Neural Network": (
        MLPClassifier(
            hidden_layer_sizes=(64, 32),
            max_iter=500, early_stopping=True,
            random_state=42
        ), True
    )
}

results = []
trained_models = {}
probabilities_by_model = {}
predictions_by_model = {}

for name, (model, use_scaling) in models.items():
    train_data = X_train_scaled if use_scaling else X_train
    test_data = X_test_scaled if use_scaling else X_test

    model.fit(train_data, y_train)

    predictions = model.predict(test_data)
    probabilities = model.predict_proba(test_data)[:, 1]

    trained_models[name] = {
        "model": model,
        "use_scaling": use_scaling
    }
    predictions_by_model[name] = predictions
    probabilities_by_model[name] = probabilities

    results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, predictions),
        "Precision": precision_score(
            y_test, predictions, zero_division=0
        ),
        "Recall": recall_score(
            y_test, predictions, zero_division=0
        ),
        "F1 Score": f1_score(
            y_test, predictions, zero_division=0
        ),
        "ROC AUC": roc_auc_score(y_test, probabilities)
    })

results_df = pd.DataFrame(results).sort_values(
    "F1 Score", ascending=False
).reset_index(drop=True)

display(results_df.round(4))
```

```python

plt.figure(figsize=(9, 5))
plot_data = results_df.melt(
    id_vars="Model",
    value_vars=["Precision", "Recall", "F1 Score", "ROC AUC"],
    var_name="Metric",
    value_name="Score"
)

sns.barplot(data=plot_data, x="Model", y="Score", hue="Metric")
plt.ylim(0, 1)
plt.title("Classification Model Comparison")
plt.xticks(rotation=25, ha="right")
plt.tight_layout()
plt.show()

best_name = results_df.iloc[0]["Model"]
best_model = trained_models[best_name]["model"]
best_scaling = trained_models[best_name]["use_scaling"]

best_predictions = predictions_by_model[best_name]
best_probabilities = probabilities_by_model[best_name]

print("Best model by F1 Score:", best_name)
print("\nClassification report:")
print(classification_report(
    y_test, best_predictions,
    target_names=["Stayed", "Left"],
    zero_division=0
))

cm = confusion_matrix(y_test, best_predictions)
sns.heatmap(
    cm, annot=True, fmt="d", cmap="Blues",
    xticklabels=["Stayed", "Left"],
    yticklabels=["Stayed", "Left"]
)
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title(f"Confusion Matrix — {best_name}")
plt.tight_layout()
plt.show()
```

```python

risk_df = X_test_raw.copy().reset_index(drop=True)
risk_df["Actual Attrition"] = y_test.reset_index(drop=True).map({
    0: "Stayed", 1: "Left"
})
risk_df["Attrition Probability"] = best_probabilities

risk_df["Risk Category"] = pd.cut(
    risk_df["Attrition Probability"],
    bins=[-0.001, 0.30, 0.60, 1.001],
    labels=["Low", "Medium", "High"]
)

display(
    risk_df.sort_values(
        "Attrition Probability", ascending=False
    ).head(20)
)

print("\nRisk category counts:")
display(risk_df["Risk Category"].value_counts().to_frame("Employees"))

plt.figure(figsize=(6, 4))
sns.countplot(
    data=risk_df, x="Risk Category",
    order=["Low", "Medium", "High"]
)
plt.title("Predicted Employee Risk Categories")
plt.tight_layout()
plt.show()

if "Department" in risk_df.columns:
    print("\nDepartment-wise average predicted risk:")
    display(
        risk_df.groupby("Department", observed=False)
        ["Attrition Probability"].mean()
        .mul(100).sort_values(ascending=False)
        .rename("Average Predicted Risk (%)").to_frame()
    )
```

```python

total_employees = len(df)
employees_left = int(df["Attrition"].sum())
employees_retained = total_employees - employees_left

print("Total employees:", total_employees)
print("Employees who left:", employees_left)
print("Employees retained:", employees_retained)
print(f"Attrition rate: {employees_left / total_employees:.2%}")
print(f"Retention rate: {employees_retained / total_employees:.2%}")

if "Department" in df.columns:
    dept_kpis = df.groupby("Department").agg(
        Total_Employees=("Attrition", "size"),
        Employees_Left=("Attrition", "sum"),
        Attrition_Rate=("Attrition", "mean")
    )
    dept_kpis["Retention_Rate"] = 1 - dept_kpis["Attrition_Rate"]
    dept_kpis["Attrition_Rate"] *= 100
    dept_kpis["Retention_Rate"] *= 100
    display(dept_kpis.round(2))

results_df.to_csv("/content/model_comparison.csv", index=False)
risk_df.to_csv("/content/employee_attrition_predictions.csv", index=False)
df.to_csv("/content/cleaned_hr_data.csv", index=False)
```

```python

%%writefile /content/api.py
from flask import Flask, request, jsonify
import pandas as pd
import numpy as np
import joblib

app = Flask(__name__)
artifact = joblib.load("/content/best_attrition_model.joblib")

def prepare_input(record):
    raw = pd.DataFrame([record])

    expected = artifact["raw_columns"]
    raw = raw.reindex(columns=expected)

    encoded = pd.get_dummies(raw, drop_first=True, dtype=int)
    encoded = encoded.reindex(
        columns=artifact["feature_columns"], fill_value=0
    )

    medians = pd.Series(artifact["medians"])
    encoded = encoded.fillna(medians).fillna(0)

    if artifact["use_scaling"]:
        return artifact["scaler"].transform(encoded)

    return encoded

@app.route("/", methods=["GET"])
def home():
    return jsonify({
        "message": "Employee Attrition Prediction API",
        "endpoint": "/predict",
        "method": "POST"
    })

@app.route("/predict", methods=["POST"])
def predict():
    record = request.get_json(silent=True)

    if not isinstance(record, dict):
        return jsonify({"error": "Send a JSON employee record."}), 400

    try:
        X = prepare_input(record)
        model = artifact["model"]
        probability = float(model.predict_proba(X)[0, 1])

        if probability < 0.30:
            risk = "Low"
        elif probability < 0.60:
            risk = "Medium"
        else:
            risk = "High"

        return jsonify({
            "model": artifact["model_name"],
            "attrition_probability": round(probability, 4),
            "predicted_attrition": "Yes" if probability >= 0.5 else "No",
            "risk_category": risk
        })
    except Exception as exc:
        return jsonify({"error": str(exc)}), 400

if __name__ == "__main__":
    app.run(host="0.0.0.0", port=5000, debug=False)
```

```python

%%writefile /content/dashboard.py
import streamlit as st
import pandas as pd
import joblib
import matplotlib.pyplot as plt

st.set_page_config(page_title="Employee Attrition Analytics", layout="wide")
st.title("Employee Attrition Analytics Platform")

df = pd.read_csv("/content/cleaned_hr_data.csv")
results = pd.read_csv("/content/model_comparison.csv")
artifact = joblib.load("/content/best_attrition_model.joblib")

total = len(df)
left = int(df["Attrition"].sum())
retained = total - left

c1, c2, c3, c4 = st.columns(4)
c1.metric("Total Employees", total)
c2.metric("Employees Left", left)
c3.metric("Attrition Rate", f"{left / total:.2%}")
c4.metric("Retention Rate", f"{retained / total:.2%}")

st.subheader("Attrition Distribution")
fig, ax = plt.subplots()
df["Attrition"].map({0: "Stayed", 1: "Left"}).value_counts().plot.bar(ax=ax)
ax.set_ylabel("Employees")
st.pyplot(fig)

if "Department" in df.columns:
    st.subheader("Department-wise Attrition")
    dept = df.groupby("Department")["Attrition"].mean().mul(100)
    st.bar_chart(dept)

st.subheader("Model Performance")
st.dataframe(results, use_container_width=True)
st.write("Selected model:", artifact["model_name"])

st.subheader("Predict Employee Attrition")
st.write("Upload a CSV with the original employee feature columns.")

upload = st.file_uploader("Upload employee CSV", type=["csv"])

if upload is not None:
    raw = pd.read_csv(upload)
    if "Attrition" in raw.columns:
        raw = raw.drop(columns=["Attrition"])

    raw = raw.reindex(columns=artifact["raw_columns"])
    encoded = pd.get_dummies(raw, drop_first=True, dtype=int)
    encoded = encoded.reindex(
        columns=artifact["feature_columns"], fill_value=0
    )
    encoded = encoded.fillna(pd.Series(artifact["medians"])).fillna(0)

    if artifact["use_scaling"]:
        model_input = artifact["scaler"].transform(encoded)
    else:
        model_input = encoded

    probs = artifact["model"].predict_proba(model_input)[:, 1]
    out = raw.copy()
    out["Attrition Probability"] = probs
    out["Risk Category"] = pd.cut(
        probs, [-0.001, 0.30, 0.60, 1.001],
        labels=["Low", "Medium", "High"]
    ).astype(str)

    st.dataframe(out, use_container_width=True)
    st.download_button(
        "Download predictions",
        out.to_csv(index=False),
        "attrition_predictions.csv",
        "text/csv"
    )
```
